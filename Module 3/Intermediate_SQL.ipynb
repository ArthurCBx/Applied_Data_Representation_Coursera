# Using String Patterns and Ranges

## Retrieving rows - using a String Pattern
- Example: we just know the first name of an author but we want to search it in the database
  - SELECT firstname FROM AUTHOR  WHERE firstname like 'R%'
    - like is used to search for a pattern in a column;
    - The '%' is used to define missing leters. Can be used before or after the pattern or both at the same time.

## Retrieving rows - using a Range
- Example: We want to get books where the number of pages is between 290 and 300
  - SELECT title, pages FROM BOOKS WHERE pages >= 290 AND pages <= 300.
  - SELECT title, pages FROM BOOKS WHERE pages between 290 and 300.

## Retrieving rows - using a Set of Values
- Example: We want to search books where the author are from two given countries.
  - SELECT firstname, lastname, country FROM AUTHOR WHERE country='AU' OR country='BR'.
  - SELECT firstname, lastname, country FROM AUTHOR WHERE country IN ('AU','BR')


# Sorting Result Sets

## Using the ORDER BY clause
- Example: alphabetical order (by default in ascending order);
  - SELECT title FROM BOOK ORDER BY title;
  - SELECT title FROM BOOK ORDER BY title DESC;

- Example: numerical order
  - SELECT title, pages FROM BOOK ORDER BY 2;

# Grouping Result Sets

## Eliminating Duplicates - DISTINCT clause
- Example: To see how many authors come from the same country.
  - SELECT country, COUNT(country) FROM AUTHOR GROUP BY country;
  - SELECT country, COUNT(country) as Count from AUTHOR GROUP BY country; -> this name the column as 'Count'.

## Restricting the Result Set - HAVING clause
- Example: To see how many authors come from the same country but only display if this number is greater than 4.
```sql
SELECT country, COUNT(country)
AS Count FROM AUTHOR
GROUP BY country HAVING COUNT(country) > 4;
```

# Buit-in Database Functions
- Most databases come with built-in SQL functions;
- Built-in functions can be included as part of SQL statements;
- Database functions can significantly reduce the amount of data that needs to be retrieved;
- Can speed up data processing;

## Aggregate or Column Functions
- Input: Collection of values;
- Output: Single value;
- Examples: SUM(), MIN(), MAX(), AVG() etc

### SUM
- Add up all the values in a column
  - Example: SELECT SUM(COST) FROM table_name;

### MIN, MAX
- MIN: Return the minimum value;
- Max: Return the maximum value.

### AVG
- Return the average value.

## SCALAR and STRING Functions
### Scalar
- Perform operations on every input value
- Examples: ROUND(), LENGTH(), UPPER(), LOWER()


# Date and Time Built-in Functions
- Most databases contain special datatypes for dates and times:
  - DATE: YYYYMMDD
  - TIME: HHMMSS
  - TIMESTAMP: YYYYXXDDHHMMSSZZZZZZ

- Date time functions:
  - YEAR(), MONTH(), DAY(), DAYOFMONTH(), DAYOFWEEK(), DAYOFYEAR(), WEEK(), HOUR(), MINUTE(), SECOND().

## Date or Time Arithmetic:
  - What date is it 3 days after each rescue date?
    - SELECT DATE_ADD(RESCUEDATE, INTERVAL 3 DAY) FROM PETRESCUE;
  
  - Special Registers:
    - CURRENT_DATE, CURRENT_TIME
  
  - How many days have passed since each RESCUEDATE till now:
    - SELECT FROM_DAYS(DATEDIFF(CURRENT_DATE, RESCUEDATE)) FROM PETRESCUE

# Sub-Queries and Nested Selects
- Sub-query: A query inside another query
  - Example:
    ```sql
    SELECT column1 FROM TABLE
    WHERE column2 = (SELECT MAX(column2) FROM TABLE)
    ```
- Why use sub-queries?
  - Because the key in aggregate functions can sometimes not be evaluated in the where clause;
  - Example -> to retrieve the list of employees who earn more than the average salary:
    - Cannot evaluate AVG() in the WHERE clause;
    ```sql
    SELECT EMP_ID, F_NAME, L_NAME, SALARY
    FROM EMPLOYEES
    WHERE SALARY <
    (SELECT AVG(SALARY) FROM EMPLOYEES);
    ```
## Sub-queries in list of columns
- Substitute column name with a sub-query;
- Called Column Expressions
- Example: To compare the salary of each employee with the average salary
  ```sql
  SELECT EMP_ID, SALARY,
  ( SELECT AVG(SALARY) FROM EMPLOYEES ) AS AVG_SALARY
  FROM EMPLOYEES;
  ```

## Sub-queries in FROM clause
- Substitute the table name with a sub-query;
- Called Derived Tables or Table Expressions;
- Example: To retrieve non-sensitive information from the table (not including salary, birthdate for example)
  ```sql
  SELECT * FROM
  (SELECT EMP_ID, F_NAME, L_NAME, DEP_ID FROM EMPLOYEES) AS EMP4ALL;
  ```


# Working with Multiple Tables
- Ways to access multiple tables in the same query:
1. Sub-queries;
2. Implicit JOIN;
3. JOIN operators (INNER, OUTER, LEFT ...)

## Accessing multiple tables with sub-queries
- Example to retrieve employees only if their department id is in the table called departments:
```sql
SELECT * FROM EMPLOYEES
WHERE DEP_ID IN
(SELECT DEPT_ID_DEP FROM DEPARTMENTS);
```
- Example to retrieve information of employees if the employee salary is greater than 70000
```sql
SELECT DEPT_ID_DEP, DEP_NAME FROM DEPARTMENTS
WHERE DEPT_ID_DEP IN
(SELECT DEP_ID FROM EMPLOYEES WHERE SALARY > 70000)
```

## Accessing multiple tables with Implicit Join
- Specfifying 2 tables in the FROM clause results in a full join:
```sql
SELECT * FROM EMPLOYEES, DEPARTMENTS;
```
- Limiting the result set to only rows with matching id:
```sql
SELECT * FROM EMPLOYEES E, DEPARTMENTS D
WHERE E.DEP_ID = D.DEPT_ID_DEP;
```
