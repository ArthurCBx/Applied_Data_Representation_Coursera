# Overview of Common Linux Shell Commands

## What is a shell ?
- User interface for running commands;
- Interactive language;
- Scripting language to automate tasks;
### Applications
- Getting information;
- Navigating and working with files and directories;
- Printing file and string contents;
- Performing network operations;
- Monitoring performance and status;
- Running batch jobs.
## A sea of shells
- Default: Bash
- Others: sh, ksh, tcsh, zsh, fish.

## Getting information
- Some common shell commands for getting information include:
  - whoami - username;
  - id - user ID and group ID;
  - uname - operating system name;
  - ps - running processes;
  - top - resource usage;
  - df - mounted file systems;
  - man - reference manual;
  - date - today's date.

## Working with files
- Some common shell commands for working with files include:
- cp - copy file;
- mv - change file name or path;
- rm - remove file;
- touch - create empty file, update file timestamp;
- chmod - change/modify file permissions;
- wc - get count of lines, words, characters in file;
- grep - return lines in file matching pattern.

## Navigating and working with directories
- Common shell commands for navigating and working with directories include:
  - ls - list files and directories;
  - find - find files in directory tree;
  - pwd - get present working directory;
  - mkdir - make directory;
  - cd - change directory;
  - rmdir - remove directory.

## Printing file and string contents
- For printing file contents or strings, common commands include:
  - cat - print file contents;
  - more - print file contents page-by-page;
  - head - print first N lines of file;
  - tail - print last N lines of file;
  - echo - print string or variable value.

## Compression and archiving
- Shell commands related to file compression and archiving applications include:
  - tar - archive a set of files;
  - zip - compress a set of files;
  - unzip - extract files from a compressed zip archive.

## Networking
- Networking applications include:
  - hostname - print hostname;
  - ping - send packets to URL and print response;
  - ifconfig - display or configure system network interfaces;
  - curl - display contents of file at a URL;
  - wget - download file from a URL.

## Running Linux on a Windows machine
- Dual boot with a partition;
- Install Linux on a virtual machine;
- Use a Linux Emulator;
- Windows Subsystem for Linux (WSL).

# Informational Commands

## User information
- Display user information;

## System Information
- uname (Unix name) - returns OS information
  - flag -s -r: OS name and version
  - flag -v: detailed version information

## Displaying your disk usage
- df (disk free) - shows disk usage
  - flag -h ~: usage in home directory.

- ps (process status) - Monitor or manage running processes
  - flag -e: List all processes running regardless of users.

## Monitoring system health and status
- top (table of processes) - Monitor system performance and resource usage
  - flag -n N: shows the N most CPU consuming processes.

## Getting date information
- date - Displays system date and time
  - displays formatted date:
  ```bash
  "date +%j day of %Y"  
  ```

## Viewing the manual
- man (manual) - Shows manual for any command
  - flag `-k .` - lists all commands that have a manual page

# Useful Commands for Wrangling Text Files
- Sorting your views line-by-line
  - sort - Sort lines in a file
    - flag -r to reverse the order

- Excluding repeated lines from views
  - uniq("unique") - Filter out repeated lines

- Extracting lines matching a pattern
  - grep("global regular expression print") - Returns lines in file matching pattern. Example to return the lines with ch:
  ```bash
  grep ch people.txt
  ```
  flag -i to be case insensitive

- Extracting slices from lines
  - cut - Extracts a section from each line. Example to return 2nd to 9th character:
  ```bash
  cut -c 2-9 people.txt
  ```
  Example to return the 2nd name:
  ```bash
  cut -d ' ' -f2 people.txt
  ```
    - `d ' '` indicates delimiter by ' '.
    - `-f2` returns the 2nd object

- Merging lines from multiple files
  - paste - merge lines from different files. - Example: `paste first.txt last.txt yob.txt`
  - Example comma separate merge: `paste -d "," first.txt last.txt yob.txt`

# File Archiving and Compression Commands

## Archiving and compression
- Archives:
  - Store rarely used information and preserve it;
  - Are a collection of data files and directories stored as a single file;
  - Make the collection more portable and serve as a backup in case of loss or corruption.

- File Compression:
  - Reduces file size by reducing information redundancy;
  - Preserves storage space, speeds up data transfer and reduce bandwidth load

## Directory tree archiving and compressing
- ls -R: lists all files inside a directory and its subdirectories
- tar (tape archiver) - Archive and extract files. Example to archive files and subdirectories:
  ```bash
  tar -cf notes.tar notes
  ```
    - -c to create notes.tar and -f to interpret is inputs from file
  ```bash
  tar -czf notes.tar.gz notes
  ```
    - -z to compress the files

- To read from archived files:
  ```bash
  tar -tf notes.tar
  ```
- To extract files archived:
  ```bash
  tar -xf notes.tar notes
  ```
- To decompress and extract:
  ```bash
  tar -xzf notes.tar.gz notes
  ```
## File compression and archiving
- zip - Compress files and directories to an archive
  - Obs:
    - zip: Compress --> bundle
    - tar: bundle --> compress
- Example:
  ```bash
  zip -r notes.zip notes
  ```
- To unzip:
  ```bash
  unzip notes.zip
  ```
